```python
import os
import pandas as pd
import numpy as np
from PIL import Image

raw_dir = os.path.join("..", "data", "raw", "WoundTissue")
img_dir = os.path.join(raw_dir, "image")
lbl_dir = os.path.join(raw_dir, "label")

img_files = sorted([f for f in os.listdir(img_dir) if f.endswith(".png")])
lbl_files = sorted([f for f in os.listdir(lbl_dir) if f.endswith(".png")])

print(f"Total images: {len(img_files)}")
print(f"Total labels: {len(lbl_files)}")
assert img_files == lbl_files, "Mismatch between image and label filenames!"
print("100% 1-to-1 pairing verified.")
```

## 3. Findings & Patient ID Disclosure
- **Sample size**: 13 paired images in current public release (filenames: `img (1).png` to `img (31).png`).
- **Channels**: Images are 3-channel RGB. Ground truth masks are 4-channel RGBA (alpha=255).
- **Classes detected**: Background (Blue), Granulation (Red), Slough (Yellow), Necrosis (Black), Maceration (White), Bone (Cream).
- **Patient / Wound IDs**: Filenames are arbitrary sequential indices (`img (n).png`) with no patient or anatomical site metadata. Split strategies must disclose this absence to prevent false patient-grouping claims.
